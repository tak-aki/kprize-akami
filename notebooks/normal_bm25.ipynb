```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def plot_metric_distribution(df):
    fig, ax = plt.subplots(1, 3, figsize=(12, 4))
    df["precision"].hist(bins=np.arange(0, 1.2, 0.1), ax=ax[0])
    ax[0].set_title("Precision")
    df["recall"].hist(bins=np.arange(0, 1.2, 0.1), ax=ax[1])
    ax[1].set_title("Recall")
    df["map_30"].hist(bins=np.arange(0, 1.2, 0.1), ax=ax[2])
    ax[2].set_title("MAP@30")
    plt.show()
```

```python
normal_bm25_result = pd.read_csv('../output/20250310034209/result.csv')
plot_metric_distribution(normal_bm25_result)
```
